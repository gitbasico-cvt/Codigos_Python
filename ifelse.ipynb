```python
usuario = "pedro"

if usuario == "mariana":
    print("Usuário reconhecido!")
else:
    print("Usuário não reconhecido!")
```

```python
usuario = "pedro"
if usuario == "mariana":
    print("Bem-vinda, Mariana!")
elif usuario == "pedro":
    print("Bem-vindo, Pedro!")
else:
    print("Usuário não reconhecido!")
```
