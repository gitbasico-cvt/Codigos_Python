```python
# Lista inicial de produtos
produtos = ["Teclado", "Mouse", "Monitor"]
print("=== ESTOQUE FÁCIL ===")
print("1 - Listar produtos")
print("2 - Cadastrar produto")
print("3 - Atualizar produto")
print("4 - Excluir produto")
print("0 - Sair")
opcao = input("Escolha uma opção: ")
while opcao != "0":
#----------------------------------------------------------------------
  # LISTAR PRODUTOS
  if opcao == "1":
      print("Produtos Cadastrados", produtos)
  #----------------------------------------------------------------------
  # CADASTRAR PRODUTO
  elif opcao == "2":
      novo_produto = input("Digite o nome do produto: ").capitalize()
      produtos.append(novo_produto)
      print("Produto cadastrado com sucesso!")
      print(produtos)
  #----------------------------------------------------------------------
  # ATUALIZAR PRODUTO
  # Atualizar produto pelo nome usando index()
  elif opcao == "3":
      produto_antigo = input("Qual produto deseja atualizar? ").capitalize()
      if produto_antigo in produtos:
          novo_produto = input("Digite o novo nome: ").capitalize()
          posicao = produtos.index(produto_antigo)
          produtos[posicao] = novo_produto
          print("Produto atualizado com sucesso!")
          print(produtos)
      else:
          print("Produto não encontrado!")
  #----------------------------------------------------------------------
  # EXCLUIR PRODUTO
  elif opcao == "4":
      produto_excluir = input("Qual produto deseja excluir? ").capitalize()
      if produto_excluir in produtos:
          produtos.remove(produto_excluir)
          print("Produto excluído com sucesso!")
          print(produtos)
      else:
          print("Produto não encontrado!")

  else:
      print("Opção inválida!")

  # Atualiza a opção para evitar loop infinito
  opcao = input("Escolha uma opção: ")
```
